## Checking for newline, quotes, commas in Product Name for smooth SQL transition

```python
data[data["Product Name"].str.contains("\n", na=False)]["Product Name"].head()
```

```python
data[data["Product Name"].str.contains('"', regex=False, na=False)]["Product Name"].head(10)
```

```python
data[data["Product Name"].str.contains(",", regex=False, na=False)].shape
```

```python
data["Product Name"] = (
    data["Product Name"]
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
)
```
